```python
model_path = "model"
print(f"Loading model from {model_path}...")
loaded_model = mlflow.pyfunc.load_model(model_path)
print("Model loaded successfully.")

proportions = []

for index, row in resultados_modelos.iterrows():
    version_modelo = row['test_name']
    file_path = f'resumenes/resultados_{version_modelo}.csv'
    col_name = f'pls_{version_modelo}'
    print(f'Analizando modelo {version_modelo}')

    data = pd.read_csv(file_path, sep=';')
        
    if col_name in data.columns:
        texts_to_predict = data[col_name].dropna().astype(str).tolist()
        
        if len(texts_to_predict) > 0:
            preds = loaded_model.predict(texts_to_predict)
            prop_1 = preds.mean() 
            proportions.append(prop_1)
        else:
            proportions.append(0.0) # No text data found
    else:
        print(f"❌ Column '{col_name}' not found in {file_path}")
        proportions.append(None)

resultados_modelos['clasificador_pls'] = proportions

print("\nProcessing complete. First 5 rows:")
print(resultados_modelos[['test_name', 'clasificador_pls']].head())
```

```python
resultados_modelos.to_excel('metricas_clasificador_modelos.xlsx', index=False)
```
