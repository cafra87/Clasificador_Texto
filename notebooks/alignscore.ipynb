```python
from alignscore import AlignScore
import pandas as pd
from tqdm.auto import tqdm
```

```python
scorer = AlignScore(model='roberta-large'
                    , batch_size=512
                    , device='cpu'
                    , ckpt_path='../../AlignScore/AlignScore-large.ckpt'
                    , evaluation_mode='nli_sp')
```

```python
version_modelo = "Qwen3-0.6B-01"
```

```python
REFERENCE_COL = 'pls' 
prediction_col = f'pls_{version_modelo}'
```

```python
data = pd.read_csv(f'../datos/resumenes_generados/resultados_{version_modelo}.csv', sep=';')
```

```python
data[f'alignscore'] = scorer.score(contexts=data[REFERENCE_COL], claims=data[prediction_col])
```

```python
data.to_csv(f'../datos/resultados_metricas/alignscore_{version_modelo}.csv', sep=';', index=False)
```
